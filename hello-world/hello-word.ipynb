```python
import ollama
import json

# No extra libraries needed!
data_to_store = []

# 1. Get the numbers from the local server
response = ollama.embed(model='nomic-embed-text', input='Setting up Intel GPU drivers')
vector = response['embeddings'][0]

# 2. Package it up
entry = {
    "text": "Setting up Intel GPU drivers",
    "vector": vector
}
data_to_store.append(entry)

# 3. Save it to disk using the built-in JSON module
with open("my_vectors.json", "w") as f:
    json.dump(data_to_store, f)

print("Saved 1 vector to my_vectors.json without any heavy dependencies!")
```

```python
import ollama
import json
import math

# --- Helper Functions ---

def dot_product(v1, v2):
    return sum(a * b for a, b in zip(v1, v2))

def magnitude(v):
    return math.sqrt(sum(a * a for a in v))

def cosine_similarity(v1, v2):
    return dot_product(v1, v2) / (magnitude(v1) * magnitude(v2))

# --- 1. Load your saved vectors ---
with open("my_vectors.json", "r") as f:
    database = json.load(f)

# --- 2. Embed the search query ---
query = "Intel GPU drivers?"
response = ollama.embed(model='nomic-embed-text', input=query)
query_vector = response['embeddings'][0]

# --- 3. Compare against every entry in the database ---
results = []
for entry in database:
    score = cosine_similarity(query_vector, entry["vector"])
    results.append((score, entry["text"]))

# --- 4. Sort and show the best match ---
results.sort(reverse=True)
print(f"Query: '{query}'")
print(f"Best match: '{results[0][1]}' (score: {results[0][0]:.4f})")
```
